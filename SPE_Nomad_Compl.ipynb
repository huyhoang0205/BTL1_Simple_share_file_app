### 3. Select Metrics

Trong mô hình hàng đợi của hệ thống buffet, các chỉ số đánh giá hiệu suất (performance metrics)
được chọn bao gồm:

#### **Response Time Metrics**
- **Wq** – Thời gian chờ trung bình trong hàng đợi (phút)
- **W** – Thời gian trung bình trong hệ thống (chờ + phục vụ)
- **System Time** – Tổng thời gian từ lúc khách vào hệ thống đến khi rời hệ thống  

#### **Throughput Metrics**
- **λ_eff** – Tốc độ đến hiệu dụng (khách được phục vụ/giờ)  
- **Throughput** – Số khách hoàn tất trung bình mỗi giờ
- **ρ (Utilization)** – Tỷ lệ bận của server trên tổng thời gian (0–1)  

#### **Queue Metrics**
- **Lq** – Số khách trung bình trong hàng đợi  
- **L** – Số khách trung bình trong toàn hệ thống
- **Pb** – Xác suất bị chặn (blocking probability)  

#### **System-Level Metrics**
- **System Efficiency** – Served / (Served + Blocked)
- **Customer Satisfaction** – Đánh giá dựa trên thời gian chờ (≤ 5 phút được xem là tốt)

#### **Validation Metrics**
- **Little's Law:**  
  \( L = \lambda \times W \)
- **Theoretical M/M/c/K Comparison** – Compare simulation results with theoretical queueing model
- **Error Percentage:**  
  \( \frac{\text{Simulation} - \text{Theory}}{\text{Theory}} \)

## CUSTOMER

Đại diện cho một khách hàng trong hệ thống buffet


````python
from typing import List, Dict

class Customer:
    def __init__(self, customer_id: str, entrance: str, arrival_time: float):
        """
        Khởi tạo Customer
        """
        self.id = customer_id
        self.arrival_time = arrival_time
        self.entrance = entrance
        self.current_station = None
        self.stations_visited: List[str] = []
        self.waiting_times: Dict[str, float] = {}
        self.service_times: Dict[str, float] = {}
        self.exit_time = None

    def visit_station(self, station: str):
        """
        Ghi nhận khách đến thăm một station
        """
        self.current_station = station
        if station not in self.stations_visited:
            self.stations_visited.append(station)

    def record_wait_time(self, station: str, time: float):

        self.waiting_times[station] = time

    def record_service_time(self, station: str, time: float):

        self.service_times[station] = time

    def get_total_time(self) -> float:

        if self.exit_time:
            return self.exit_time - self.arrival_time
        return 0.0

    def get_journey(self) -> List[str]:

        return self.stations_visited

    def __repr__(self):
        return f"Customer(id={self.id}, entrance={self.entrance})"
````

## ENTRANCE

Entrance Module
Định nghĩa class Entrance - điểm vào của khách hàng

````python
import simpy
import numpy as np
from typing import Dict


class Entrance:
    def __init__(self, name: str, env: simpy.Environment, arrival_rate: float,
                 routing_probs: Dict[str, float], buffet_system):
        """
        Khởi tạo Entrance

        Args:
            name: Tên entrance (Entrance1, Entrance2)
            env: SimPy environment
            arrival_rate: Arrival rate λ (customers/hour)
            routing_probs: Xác suất routing đến các station đầu tiên
            buffet_system: Reference đến BuffetSystem
        """
        self.name = name
        self.env = env
        self.arrival_rate = arrival_rate
        self.routing_probs = routing_probs
        self.buffet_system = buffet_system
        self.customer_count = 0

    def generate_arrivals(self):
        while True:
            # Interarrival time theo Exponential
            interarrival_time = self.get_interarrival_time()
            yield self.env.timeout(interarrival_time)

            # Tạo customer mới
            customer = self.create_customer()

            # Chọn station đầu tiên
            first_station = self.route_new_customer()

            # Bắt đầu journey của customer
            self.env.process(
                self.buffet_system.customer_journey(customer, first_station)
            )

    def get_interarrival_time(self) -> float:
        """Tạo interarrival time theo Exponential distribution """
        return np.random.exponential(1.0 / self.arrival_rate)

    def create_customer(self) -> Customer:
        """ Tạo customer mới với ID duy nhất"""
        self.customer_count += 1
        customer_id = f"{self.name}_C{self.customer_count}"
        return Customer(customer_id, self.name, self.env.now)

    def route_new_customer(self) -> str:
        """ Chọn station đầu tiên """
        stations = list(self.routing_probs.keys())
        probabilities = list(self.routing_probs.values())
        return np.random.choice(stations, p=probabilities)

    def __repr__(self):
        return f"Entrance({self.name}, λ={self.arrival_rate})"

````

## ROUTER

````python

import numpy as np
from typing import Dict


class Router:
    def __init__(self, routing_matrix: Dict[str, Dict[str, float]],
                 stations: Dict,
                 balking_threshold: float = 0.7):
        self.routing_matrix = routing_matrix
        self.stations = stations
        self.balking_threshold = balking_threshold

    def get_next_station(self, current: str, customer) -> str:
        if current not in self.routing_matrix:
            return 'Exit'

        probs = self.routing_matrix[current]
        stations = list(probs.keys())
        probabilities = list(probs.values())

        next_station = np.random.choice(stations, p=probabilities)

        # Kiểm tra blocking và balking
        if next_station != 'Exit' and next_station in self.stations:
            station = self.stations[next_station]

            # Blocking: queue đầy
            if station.is_full():
                return self.handle_blocking(current, next_station, customer)

            # Balking: queue quá dài
            if self.apply_balking(station):
                return self.handle_blocking(current, next_station, customer)

        return next_station

    def handle_blocking(self, current: str, blocked_station: str, customer) -> str:
        """ \"\"\"
        Xử lý khi station bị block
        Tìm station thay thế hoặc Exit
        """
        if current not in self.routing_matrix:
            return 'Exit'

        probs = self.routing_matrix[current].copy()
        if blocked_station in probs:
            del probs[blocked_station]

        if not probs:
            return 'Exit'

        # Normalize lại xác suất
        total = sum(probs.values())
        if total > 0:
            probs = {k: v/total for k, v in probs.items()}
            stations = list(probs.keys())
            probabilities = list(probs.values())
            return np.random.choice(stations, p=probabilities)

        return 'Exit'

    def apply_balking(self, station) -> bool:
        queue_ratio = station.get_current_queue_size() / station.capacity
        if queue_ratio > self.balking_threshold:
            # Xác suất balking tăng theo độ dài queue
            balking_prob = (queue_ratio - self.balking_threshold) / (1 - self.balking_threshold)
            return np.random.random() < balking_prob
        return False

    def update_probabilities(self, time: float):
        """ \"\"\"
        Cập nhật routing probabilities theo thời gian (giờ cao điểm)
        TODO: Implement dynamic routing

        Args:
            time: Thời gian hiện tại
        \"\"\" """
        pass
````

````python
from dataclasses import dataclass, field
from typing import List, Dict
import numpy as np
import pandas as pd


@dataclass
class StationStatistics:
    """ \"\"\"Thống kê cho một station cụ thể\"\"\" """

    waiting_times: List[float] = field(default_factory=list)
    service_times: List[float] = field(default_factory=list)
    queue_lengths: List[int] = field(default_factory=list)
    queue_length_times: List[float] = field(default_factory=list)
    customers_served: int = 0
    customers_blocked: int = 0
    total_busy_time: float = 0.0

    def record_waiting_time(self, time: float):
        """ \"\"\"Ghi nhận waiting time\"\"\" """
        self.waiting_times.append(time)

    def record_service_time(self, time: float):

        self.service_times.append(time)

    def record_queue_length(self, length: int, time: float):
        """ Ghi nhận độ dài queue tại thời điểm time"""
        self.queue_lengths.append(length)
        self.queue_length_times.append(time)

    def get_avg_waiting_time(self) -> float:

        return np.mean(self.waiting_times) if self.waiting_times else 0.0

    def get_avg_service_time(self) -> float:

        return np.mean(self.service_times) if self.service_times else 0.0

    def get_avg_queue_length(self) -> float:

        return np.mean(self.queue_lengths) if self.queue_lengths else 0.0

    def get_blocking_probability(self) -> float:

        total = self.customers_served + self.customers_blocked
        return self.customers_blocked / total if total > 0 else 0.0


@dataclass
class SystemStatistics:


    station_stats: Dict[str, StationStatistics] = field(default_factory=dict)
    customer_journeys: List = field(default_factory=list)
    total_throughput: int = 0
    system_times: List[float] = field(default_factory=list)

    def collect_customer_data(self, customer):

        self.customer_journeys.append(customer)
        self.system_times.append(customer.get_total_time())
        self.total_throughput += 1

    def get_system_summary(self) -> dict:

        summary = {
            'total_customers': self.total_throughput,
            'avg_system_time': np.mean(self.system_times) if self.system_times else 0,
            'stations': {}
        }

        for name, stats in self.station_stats.items():
            summary['stations'][name] = {
                'avg_waiting_time': stats.get_avg_waiting_time(),
                'avg_service_time': stats.get_avg_service_time(),
                'avg_queue_length': stats.get_avg_queue_length(),
                'customers_served': stats.customers_served,
                'customers_blocked': stats.customers_blocked,
                'blocking_probability': stats.get_blocking_probability()
            }

        return summary

    def export_to_dataframe(self) -> pd.DataFrame:

        data = []
        for customer in self.customer_journeys:
            for station in customer.stations_visited:
                data.append({
                    'customer_id': customer.id,
                    'entrance': customer.entrance,
                    'station': station,
                    'waiting_time': customer.waiting_times.get(station, 0),
                    'service_time': customer.service_times.get(station, 0),
                    'total_time': customer.get_total_time()
                })
        return pd.DataFrame(data)

    def compare_with_theory(self) -> dict:
        """ \"\"\"So sánh kết quả simulation với công thức toán học\"\"\" """
        # TODO: Implement theoretical calculations
        # Dùng công thức M/M/c/K để tính Lq, Wq, etc.
        return {}

````

## STATION
###Đại diện cho một quầy phục vụ (station) trong buffet
####Mô hình: M/M/c/K


    * M: Poisson arrival
    * M: Exponential service
    * c: số server
    * K: capacity (max customers in system)

````python
import simpy
import numpy as np



class FoodStation:
    def __init__(self, name: str, env: simpy.Environment, num_servers: int,
                 capacity: int, service_rate: float):
        """
            name: Tên station (Salad, Appetizer, Main, Dessert, Beverage)
            env: SimPy environment
            num_servers: Số lượng server (c trong M/M/c/K)
            capacity: Capacity tối đa (K trong M/M/c/K)
            service_rate: Service rate μ (customers/hour)
        """
        self.name = name
        self.env = env
        self.num_servers = num_servers
        self.capacity = capacity
        self.service_rate = service_rate
        self.servers = simpy.Resource(env, capacity=num_servers)
        self.queue_size = 0
        self.stats = StationStatistics()

    def serve_customer(self, customer):
        arrival_at_station = self.env.now
        # Kiểm tra capacity (blocking)
        if self.is_full():
            self.stats.customers_blocked += 1
            return False
        # Tăng queue size
        self.queue_size += 1
        self.stats.record_queue_length(self.queue_size, self.env.now)
        # Request server
        with self.servers.request() as request:
            yield request

            # Ghi nhận waiting time
            wait_time = self.env.now - arrival_at_station
            self.stats.record_waiting_time(wait_time)
            customer.record_wait_time(self.name, wait_time)

            # Service time (Exponential distribution)
            service_time = self.get_service_time()
            yield self.env.timeout(service_time)

            # Ghi nhận service time
            self.stats.record_service_time(service_time)
            customer.record_service_time(self.name, service_time)
            self.stats.customers_served += 1

            # Giảm queue size
            self.queue_size -= 1
            self.stats.record_queue_length(self.queue_size, self.env.now)

        return True

    def get_service_time(self) -> float:
        return np.random.exponential(1.0 / self.service_rate)

    def is_full(self) -> bool:
        """ kiểm tra xem station có đầy capacity không """
        return self.queue_size >= self.capacity

    def get_current_queue_size(self) -> int:
        return self.queue_size

    def get_utilization(self) -> float:
        """
        Tính utilization ρ = λ/(c*μ)
        """
        if not self.stats.service_times or self.env.now == 0:
            return 0.0
        total_service = sum(self.stats.service_times)
        return total_service / (self.num_servers * self.env.now)

    def __repr__(self):
        return f"FoodStation({self.name}, c={self.num_servers}, K={self.capacity})"

````

## BUFFET_SYSTEM

````python
import simpy
from typing import Dict, List
import matplotlib.pyplot as plt
import numpy as np


class BuffetSystem:
    def __init__(self, config: dict):
        self.env = simpy.Environment()
        self.config = config
        self.stations: Dict[str, FoodStation] = {}
        self.entrances: List[Entrance] = []
        self.statistics = SystemStatistics()
        self.total_customers = 0
        self.simulation_time = 0
        # Setup các components
        self.setup_stations()
        self.setup_router()
        self.setup_entrances()

    def setup_stations(self):
        """Khởi tạo tất cả food stations từ config"""
        stations_config = self.config['stations']

        for name, params in stations_config.items():
            station = FoodStation(
                name=name,
                env=self.env,
                num_servers=params['servers'],
                capacity=params['capacity'],
                service_rate=params['service_rate']
            )
            self.stations[name] = station
            self.statistics.station_stats[name] = station.stats

    def setup_router(self):
        """ \"\"\"Khởi tạo router với routing matrix\"\"\" """
        self.router = Router(
            routing_matrix=self.config['routing_matrix'],
            stations=self.stations,
            balking_threshold=self.config.get('balking_threshold', 0.7)
        )

    def setup_entrances(self):
        """ \"\"\"Khởi tạo các lối vào\"\"\" """
        entrances_config = self.config['entrances']

        for name, params in entrances_config.items():
            entrance = Entrance(
                name=name,
                env=self.env,
                arrival_rate=params['arrival_rate'],
                routing_probs=params['routing_probs'],
                buffet_system=self
            )
            self.entrances.append(entrance)

    def customer_journey(self, customer: Customer, first_station: str):
        current_station = first_station

        while current_station != 'Exit':
            if current_station in self.stations:
                station = self.stations[current_station]
                customer.visit_station(current_station)
                success = yield self.env.process(station.serve_customer(customer))
                if not success:
                    current_station = self.router.handle_blocking(
                        current_station, current_station, customer
                    )
                    continue
            current_station = self.router.get_next_station(current_station, customer)
        customer.exit_time = self.env.now
        self.statistics.collect_customer_data(customer)
        self.total_customers += 1

    def run(self, until: float):
        self.simulation_time = until
        for entrance in self.entrances:
            self.env.process(entrance.generate_arrivals())
        self.env.run(until=until)

    def get_results(self) -> dict:
        return self.statistics.get_system_summary()

    def plot_results(self):
        """ \"\"\"Vẽ các biểu đồ phân tích kết quả\"\"\" """
        summary = self.get_results()
        fig, axes = plt.subplots(2, 2, figsize=(15, 10))
        fig.suptitle('Buffet System Simulation Results', fontsize=16, fontweight='bold')

        stations = list(summary['stations'].keys())

        # 1. Average Waiting Time by Station
        wait_times = [summary['stations'][s]['avg_waiting_time'] * 60 for s in stations]  # Convert to minutes

        axes[0, 0].bar(stations, wait_times, color='skyblue', edgecolor='navy', alpha=0.7)
        axes[0, 0].set_title('Average Waiting Time by Station', fontweight='bold')
        axes[0, 0].set_ylabel('Time (minutes)')
        axes[0, 0].tick_params(axis='x', rotation=0)
        axes[0, 0].grid(axis='y', alpha=0.3)

        # 2. Average Queue Length by Station
        queue_lengths = [summary['stations'][s]['avg_queue_length'] for s in stations]

        axes[0, 1].bar(stations, queue_lengths, color='lightcoral', edgecolor='darkred', alpha=0.7)
        axes[0, 1].set_title('Average Queue Length by Station', fontweight='bold')
        axes[0, 1].set_ylabel('Number of Customers')
        axes[0, 1].tick_params(axis='x', rotation=0)
        axes[0, 1].grid(axis='y', alpha=0.3)

        # 3. Customers Served vs Blocked
        served = [summary['stations'][s]['customers_served'] for s in stations]
        blocked = [summary['stations'][s]['customers_blocked'] for s in stations]

        x = np.arange(len(stations))
        width = 0.35

        axes[1, 0].bar(x - width/2, served, width, label='Served', color='lightgreen', edgecolor='darkgreen')
        axes[1, 0].bar(x + width/2, blocked, width, label='Blocked', color='salmon', edgecolor='darkred')
        axes[1, 0].set_title('Customers Served vs Blocked', fontweight='bold')
        axes[1, 0].set_ylabel('Number of Customers')
        axes[1, 0].set_xticks(x)
        axes[1, 0].set_xticklabels(stations, rotation=0)
        axes[1, 0].legend()
        axes[1, 0].grid(axis='y', alpha=0.3)

        if self.statistics.system_times:
            system_times_minutes = [t * 60 for t in self.statistics.system_times]  # Convert to minutes
            axes[1, 1].hist(system_times_minutes, bins=30, color='plum', edgecolor='purple', alpha=0.7)
            axes[1, 1].set_title('System Time Distribution', fontweight='bold')
            axes[1, 1].set_xlabel('Time (minutes)')
            axes[1, 1].set_ylabel('Frequency')
            axes[1, 1].axvline(np.mean(system_times_minutes),
                              color='red', linestyle='--', linewidth=2, label='Mean')
            axes[1, 1].legend()
            axes[1, 1].grid(alpha=0.3)

        plt.tight_layout()
        plt.show()

        # Print summary
        self._print_summary(summary)

    def _print_summary(self, summary: dict):
        """ \"\"\"In summary report ra console\"\"\" """
        print("\\n" + "="*70)
        print(" " * 15 + "BUFFET SYSTEM SIMULATION SUMMARY")
        print("="*70)
        print(f"Simulation Time: {self.simulation_time} hours")
        print(f"Total Customers Completed Journey: {summary['total_customers']}")
        print(f"Average System Time: {summary['avg_system_time']:.4f} hours ({summary['avg_system_time']*60:.2f} minutes)")

        print("\\n" + "-"*70)
        print("STATION DETAILS:")
        print("-"*70)

        for station, stats in summary['stations'].items():
            print(f"\\n📍 {station}:")
            print(f"   ├─ Customers Served: {stats['customers_served']}")
            print(f"   ├─ Customers Blocked: {stats['customers_blocked']}")
            print(f"   ├─ Blocking Probability: {stats['blocking_probability']:.4f}")
            print(f"   ├─ Avg Waiting Time: {stats['avg_waiting_time']:.4f} hours ({stats['avg_waiting_time']*60:.2f} min)")
            print(f"   ├─ Avg Service Time: {stats['avg_service_time']:.4f} hours ({stats['avg_service_time']*60:.2f} min)")
            print(f"   └─ Avg Queue Length: {stats['avg_queue_length']:.2f} customers")

        print("\\n" + "="*70)
````

## TEST ROUTER

### Mục đích:
Đánh giá logic định tuyến khách giữa các trạm, bao gồm cả xử lý trạm đầy và cơ chế balking.
### Các test case chính:
* Khởi tạo router: kiểm tra bảng routing, danh sách trạm và ngưỡng balking.
* Chọn trạm tiếp theo: đảm bảo hàm chọn theo xác suất hoạt động đúng; có thể tạo ra tất cả hướng đi hợp lệ.
* Route đến Exit: kiểm tra rằng các trạm cuối (như Station3) luôn dẫn khách đến Exit.
* Blocking và chuyển hướng: khi trạm kế tiếp đầy, router phải loại trạm đó và chọn trạm thay thế phù hợp.
* Cơ chế Balking: khi mức tải vượt ngưỡng, khách có thể từ chối vào trạm theo xác suất; kiểm tra hoạt động có thực sự xảy ra.
* Blocking toàn bộ: nếu tất cả các trạm thay thế đều đầy, router chuyển khách trực tiếp đến Exit.
### Vai trò:
Đảm bảo luồng di chuyển khách hàng trong mô phỏng phản ánh đúng logic hệ thống khi tải thay đổi.

````python
import pytest
import simpy
import numpy as np
import sys
import unittest

class TestRouter(unittest.TestCase):
    def setUp(self):
        """ \"\"\"Setup trước mỗi test\"\"\" """
        np.random.seed(42)
        self.env = simpy.Environment()

        # Create stations
        self.stations = {
            'Station1': FoodStation('Station1', self.env, 2, 5, 10.0),
            'Station2': FoodStation('Station2', self.env, 2, 5, 10.0),
            'Station3': FoodStation('Station3', self.env, 2, 5, 10.0)
        }

        # Routing matrix
        self.routing_matrix = {
            'Station1': {'Station2': 0.5, 'Station3': 0.3, 'Exit': 0.2},
            'Station2': {'Station3': 0.6, 'Exit': 0.4},
            'Station3': {'Exit': 1.0}
        }

        self.router = Router(self.routing_matrix, self.stations, balking_threshold=0.7)

    def test_initialization(self):
        """ \"\"\"Test khởi tạo router\"\"\" """
        assert self.router.routing_matrix == self.routing_matrix
        assert len(self.router.stations) == 3
        assert self.router.balking_threshold == 0.7

    def test_get_next_station_normal(self):
        """ \"\"\"Test routing bình thường\"\"\" """
        customer = Customer("C1", "Entrance1", 0.0)
        results = []
        for _ in range(100):
            next_station = self.router.get_next_station('Station1', customer)
            results.append(next_station)
        unique_destinations = set(results)
        assert 'Station2' in unique_destinations or 'Station3' in unique_destinations or 'Exit' in unique_destinations

    def test_get_next_station_exit(self):
        """ \"\"\"Test routing đến Exit\"\"\" """
        customer = Customer("C1", "Entrance1", 0.0)
        next_station = self.router.get_next_station('Station3', customer)
        assert next_station == 'Exit'

    def test_blocking_handling(self):
        """ \"\"\"Test xử lý blocking\"\"\" """
        customer = Customer("C1", "Entrance1", 0.0)
        self.stations['Station2'].queue_size = self.stations['Station2'].capacity
        next_station = self.router.get_next_station('Station1', customer)
        assert next_station in ['Station3', 'Exit']

    def test_balking(self):
        """ \"\"\"Test balking mechanism\"\"\" """
        station = self.stations['Station2']
        station.queue_size = 0
        assert not self.router.apply_balking(station)
        station.queue_size = int(station.capacity * 0.5)
        assert not self.router.apply_balking(station)
        station.queue_size = int(station.capacity * 0.9)

        balk_count = sum(self.router.apply_balking(station) for _ in range(100))
        assert balk_count > 0

    def test_handle_blocking_with_alternatives(self):
        """ \"\"\"Test blocking với alternative stations\"\"\" """
        customer = Customer("C1", "Entrance1", 0.0)
        result = self.router.handle_blocking('Station1', 'Station2', customer)
        assert result in ['Station3', 'Exit']

unittest.main(argv=['first-arg-is-ignored'],exit=False)

````

### Mục đích:
Đảm bảo từng trạm phục vụ (FoodStation) hoạt động đúng mô hình hàng đợi và xử lý khách hàng chính xác.
### Các test case chính:

* Kiểm tra trạng thái đầy (is_full): trạm được nhận diện đầy khi đạt hoặc vượt capacity.

* Xử lý phục vụ khách: tiến trình phục vụ tăng số khách đã phục vụ, ghi nhận thời gian chờ - phục vụ.

### Vai trò:
Đảm bảo một trạm buffet đơn lẻ vận hành đúng, là nền tảng cho mô phỏng toàn hệ thống.

````python
import pytest
import simpy
import numpy as np
import sys
import unittest

class TestFoodStation(unittest.TestCase):
    """ \"\"\"Test cases cho FoodStation\"\"\" """

    def setUp(self):
        """ \"\"\"Setup trước mỗi test\"\"\" """
        np.random.seed(42)
        self.env = simpy.Environment()
        self.station = FoodStation(
            name="TestStation",
            env=self.env,
            num_servers=2,
            capacity=5,
            service_rate=10.0
        )

    def test_initialization(self):
        """ \"\"\"Test khởi tạo station\"\"\" """
        assert self.station.name == "TestStation"
        assert self.station.num_servers == 2
        assert self.station.capacity == 5
        assert self.station.service_rate == 10.0
        assert self.station.queue_size == 0

    def test_is_full(self):
        """ \"\"\"Test kiểm tra station đầy\"\"\" """
        assert not self.station.is_full()

        # Tăng queue size
        self.station.queue_size = 5
        assert self.station.is_full()

        self.station.queue_size = 6
        assert self.station.is_full()

    def test_service_time_generation(self):
        """ \"\"\"Test tạo service time\"\"\" """
        service_times = [self.station.get_service_time() for _ in range(100)]
        assert all(t > 0 for t in service_times)
        mean_time = np.mean(service_times)
        expected_mean = 1.0 / self.station.service_rate
        assert abs(mean_time - expected_mean) < 0.02  # Tolerance

    def test_serve_customer(self):
        """ \"\"\"Test serve customer process\"\"\" """
        customer = Customer("C1", "Entrance1", 0.0)

        def test_process():
            success = yield self.env.process(self.station.serve_customer(customer))
            assert success
            assert self.station.stats.customers_served == 1
            assert len(self.station.stats.service_times) == 1

        self.env.process(test_process())
        self.env.run()

    def test_blocking(self):
        """ \"\"\"Test blocking khi station đầy\"\"\" """
        self.station.queue_size = self.station.capacity

        customer = Customer("C1", "Entrance1", 0.0)

        def test_process():
            success = yield self.env.process(self.station.serve_customer(customer))
            assert not success  # Should be blocked
            assert self.station.stats.customers_blocked == 1

        self.env.process(test_process())
        self.env.run()

    def test_statistics_recording(self):
        """ \"\"\"Test ghi nhận statistics\"\"\" """
        customer1 = Customer("C1", "Entrance1", 0.0)
        customer2 = Customer("C2", "Entrance1", 0.5)

        def test_process():
            yield self.env.process(self.station.serve_customer(customer1))
            yield self.env.process(self.station.serve_customer(customer2))

        self.env.process(test_process())
        self.env.run()
        assert self.station.stats.customers_served == 2
        assert len(self.station.stats.waiting_times) == 2
        assert len(self.station.stats.service_times) == 2
        assert self.station.stats.get_avg_waiting_time() >= 0
        assert self.station.stats.get_avg_service_time() > 0


unittest.main(argv=['first-arg-is-ignored'], exit=False)
````
